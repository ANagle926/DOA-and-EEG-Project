```python
#Use the GitHub code segment to visualize .vital files

import pandas as pd
import vitaldb
import matplotlib.pyplot as plt

#caseID= patient
#track names= specific physiological parameter or drug concentration being monitored.

caseids = vitaldb.find_cases(['BIS/EEG1_WAV'])

#Read an open dataset case -> second parameter is the track name
vals = vitaldb.load_case(caseids[0], ['BIS/EEG1_WAV'] , 1/100)

#plots
plt.figure(figsize=(20,10))
plt.subplot(211)
plt.plot(vals[110000:111000], color='g')
```
