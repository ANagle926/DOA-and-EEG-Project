```python


import pandas as pd
import vitaldb
import matplotlib.pyplot as plt



#Downloads a vital file with caseid of 1 (includes all the tracks)
vf = vitaldb.VitalFile(1)
vf.to_vital('1.vital')

track_names= vitaldb.vital_trks('1.vital')
track= track_names[1]

#Downloads a specific track only (still caseid of 1)
vf_track_specific = vitaldb.VitalFile(1, track)
vf_track_specific.to_vital('track.vital')

#After loading the vital file, you can convert it to pandas or numpy format for data analysis
sample_one = vf.to_numpy(track_names[1], 1/100)
sample_two = vf_track_specific.to_numpy(track, 1/100)


plt.figure(figsize=(20, 5))
plt.plot(sample_one[:, 0])
plt.show()


plt.figure(figsize=(20, 5))
plt.plot(sample_two[:, 0])
plt.show()
```
